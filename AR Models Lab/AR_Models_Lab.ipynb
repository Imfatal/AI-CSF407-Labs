```python
from transformers import (
    AutoTokenizer,
    EncoderDecoderModel,
    AutoModelForSeq2SeqLM,
    pipeline
)
```

```python
# Load a pre-trained translation model
model_name = "google/bert2bert_L-24_wmt_en_de"

tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    pad_token="<pad>",
    eos_token="</s>",
    bos_token="<s>"
)

model = EncoderDecoderModel.from_pretrained(
    model_name,
    device_map="auto"
)

# Input sentence to translate
input_text = "Plants create energy through a process known as"

# Encode the input text
inputs = tokenizer(
    input_text,
    return_tensors="pt",
    add_special_tokens=False
).to(model.device).input_ids

# Generate the translated output
outputs = model.generate(inputs)[0]

# Decode the output
translated_text = tokenizer.decode(
    outputs,
    skip_special_tokens=True
)

print("Translated text:", translated_text)
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "google-t5/t5-base"
)

model = AutoModelForSeq2SeqLM.from_pretrained(
    "google-t5/t5-base",
    device_map="auto"
)

input_ids = tokenizer(
    "translate English to French: Let's go have some wine, cause thats how we shine",
    return_tensors="pt"
).to(model.device)

output = model.generate(
    **input_ids,
    cache_implementation="static",
    max_length=30
)

print(
    tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )
)
```

```python
model_name = "rvv-karma/English2Hinglish-Flan-T5-Base"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

input_text = "life is crazy. Not me"

input_ids = tokenizer(
    input_text,
    return_tensors="pt"
)

output_ids = model.generate(**input_ids)

output_text = tokenizer.decode(
    output_ids[0],
    skip_special_tokens=True
)

print(output_text)
```

```python
from transformers import pipeline

# Load a decoder-only GPT-2 model
generator = pipeline(
    task="text-generation",
    model="gpt2"
)

# Generate text from a prompt
output = generator(
    "The future of AI is",
    max_length=30,
    num_return_sequences=1
)

print(output[0]["generated_text"])
```

```python
# Initialize a sentiment-analysis pipeline
classifier = pipeline(
    "sentiment-analysis"
)

# Classify the sentiment
result = classifier(
    "I'd want to kick you and ensure that you are hurt"
)

print(result)
```
